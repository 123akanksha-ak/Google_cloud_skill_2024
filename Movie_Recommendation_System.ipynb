```python

import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

print("Libraries imported successfully!")
```

```python

data = {
    "title": [
        "3 Idiots",
        "Chhichhore",
        "Dangal",
        "Taare Zameen Par",
        "Zindagi Na Milegi Dobara",
        "Dil Chahta Hai",
        "War",
        "Pathaan",
        "PK",
        "Munna Bhai M.B.B.S."
    ],

    "genre": [
        "comedy drama education",
        "comedy drama college",
        "sports drama",
        "drama education family",
        "adventure comedy drama",
        "friendship comedy drama",
        "action thriller",
        "action thriller",
        "comedy drama science fiction",
        "comedy drama medical"
    ],

    "rating": [
        8.4, 8.3, 8.3, 8.4, 8.2,
        8.1, 6.5, 5.8, 8.1, 8.1
    ]
}

movies = pd.DataFrame(data)

print(movies)
```

```python

print("Total movies:", len(movies))

print("\nMovie Dataset:")
display(movies)

print("\nAverage Rating:")
print(round(movies["rating"].mean(), 2))
```

```python

popular_movies = movies.sort_values(
    by="rating",
    ascending=False
)

print("Top 5 Movies:")
display(popular_movies[["title", "rating"]].head(5))
```

```python

vectorizer = TfidfVectorizer()

genre_matrix = vectorizer.fit_transform(
    movies["genre"]
)

print("Genre processing completed!")
print("Matrix shape:", genre_matrix.shape)
```

```python

similarity_matrix = cosine_similarity(genre_matrix)

print("Similarity calculation completed!")
print("Matrix shape:", similarity_matrix.shape)
```

```python

def recommend_movies(movie_name, n=3):

    matches = movies.index[
        movies["title"].str.lower() == movie_name.lower()
    ].tolist()

    if not matches:
        print("Movie not found! Please enter a valid title.")
        return

    movie_index = matches[0]

    scores = list(enumerate(
        similarity_matrix[movie_index]
    ))

    scores = sorted(
        scores,
        key=lambda item: item[1],
        reverse=True
    )

    recommendations = [
        item for item in scores
        if item[0] != movie_index
    ][:n]

    print("Movies similar to", movie_name)

    for index, score in recommendations:
        print(
            movies.iloc[index]["title"],
            "| Similarity:",
            round(score, 2)
        )
```

```python

recommend_movies("Dangal")
recommend_movies("War")
recommend_movies("PK")
```
